```python
import pandas as pd

input_file = 'alignment_plate_python.xlsx'
output_file = 'alignment_plate_cleaned_and_formatted.xlsx'

excel_file = pd.ExcelFile(input_file)
sheet_names = excel_file.sheet_names

with pd.ExcelWriter(output_file, engine='openpyxl') as writer:
    for i in sheet_names:
        df = pd.read_excel(input_file, sheet_name=i)
        
        # 1. ลบคอลัมน์ Unnamed
        df_cleaned = df.loc[:, ~df.columns.str.contains('^Unnamed', na=False)]
        df_cleaned.columns = df_cleaned.columns.str.strip()
        
        # 2. จัดการและเปลี่ยน Format "วันที่วัด"
        for col in df_cleaned.columns:
            if 'วันที่' in col:
                # ขั้นที่ 1: แปลงข้อมูลเดิมให้เป็น Datetime ก่อน 
                # (errors='coerce' คือถ้าเจอข้อมูลที่พังหรือไม่ใช่วันที่ ให้เปลี่ยนเป็นค่าว่าง NaT จะได้ไม่ Error)
                # (dayfirst=True บังคับให้มันเข้าใจว่าตัวเลขชุดแรกคือ "วัน" ไม่ใช่ "เดือน" ป้องกันการสับสน)
                df_cleaned[col] = pd.to_datetime(df_cleaned[col], errors='coerce', dayfirst=True)
                
                # ขั้นที่ 2: แปลงกลับเป็น Text พร้อมกำหนด Format ที่ต้องการ
                # ตัวอย่างนี้คือ วัน/เดือน/ปี (ค.ศ. 4 หลัก) -> 23/01/2024
                df_cleaned[col] = df_cleaned[col].dt.strftime('%d/%m/%Y')
                
                # เปลี่ยนค่า NaT (ที่เกิดจากข้อมูลขยะ) ให้เป็นช่องว่างๆ
                df_cleaned[col] = df_cleaned[col].fillna('')
        
        # 3. เขียนลง Excel ไฟล์ใหม่
        df_cleaned.to_excel(writer, sheet_name=i, index=False)
        
        # 4. ปรับความกว้างคอลัมน์อัตโนมัติ เพื่อแก้ปัญหา ######
        worksheet = writer.sheets[i]
        for column_cells in worksheet.columns:
            length = max(len(str(cell.value)) for cell in column_cells)
            adjusted_width = length + 3
            column_letter = column_cells[0].column_letter
            worksheet.column_dimensions[column_letter].width = adjusted_width

print(f"จัด Format วันที่และสร้างไฟล์เรียบร้อย: {output_file}")
```

```python
import pandas as pd

input_file = 'alignment_plate_cleaned_and_formatted.xlsx'
output_file = 'alignment_plate_finalV2.xlsx'

excel_file = pd.ExcelFile(input_file)

with pd.ExcelWriter(output_file, engine='openpyxl') as writer:
    for i in excel_file.sheet_names:
        df = pd.read_excel(input_file, sheet_name=i)
        
        # A. ทำที่ "ชื่อคอลัมน์"
        df.columns = df.columns.str.lower().str.strip()
        
        # B. ทำที่ "ข้อมูลในทุกเซลล์" 
        # *** เปลี่ยนจาก .map เป็น .applymap ตรงนี้ครับ ***
        df = df.applymap(lambda x: x.lower().strip() if isinstance(x, str) else x)
        
        # เขียนกลับลงไฟล์
        df.to_excel(writer, sheet_name=i, index=False)
        
        # ปรับความกว้างคอลัมน์อัตโนมัติ
        worksheet = writer.sheets[i]
        for column_cells in worksheet.columns:
            length = max(len(str(cell.value)) for cell in column_cells)
            worksheet.column_dimensions[column_cells[0].column_letter].width = length + 2

print(f"ทำ Lower และ Strip เรียบร้อยแล้วที่ไฟล์: {output_file}")
```
